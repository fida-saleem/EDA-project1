```python
plt.figure(figsize=(5,4))
sns.countplot(x="Language", data=df)
plt.title("Language")
plt.xticks(rotation=45)
plt.show()
```

```python
df.hist(figsize=(12,10))
plt.tight_layout()
plt.show()
```

```python
type_count = df["Type"].value_counts()

plt.bar(type_count.index, type_count.values)

plt.title("Type")
plt.xlabel("Type")
plt.ylabel("Count")

plt.show()
```

```python
sns.scatterplot(x="Language", y="Solo", hue="Sex",
data=df)
plt.xticks(rotation=45)
plt.show()
```

```python
sns.lineplot(x="Type", y="Sex", data=df)
plt.show()
```

```python
sns.boxplot(
    x="Type",
    y="Total_Stream",
    data=df
)

plt.title("Total Streams by Type")
plt.xlabel("Type")
plt.ylabel("Total Streams")
plt.xticks(rotation=45)

plt.show()
```

```python
sns.histplot(df["Total_Stream"])

plt.title("Distribution of Total Streams")
plt.xlabel("Total Stream")
plt.ylabel("Count")

plt.show()
```

```python
sns.barplot(x="Type", y="Total_Stream", data=df)
plt.show()
```

```python
sns.boxplot(x="Total_Stream", y="Type", data=df,color="pink")
plt.show()
```

```python
# correlation of columns
plt.figure(figsize=(8,6))
sns.heatmap(df.corr(numeric_only=True),
            annot=True,
            cmap="coolwarm",
            fmt=".2f")

plt.title("Correlation Heatmap")
plt.show()
```

```python
sns.kdeplot(df["Total_Stream"])

plt.title("Distribution of Total Streams")
plt.xlabel("Total Stream")
plt.ylabel("Density")

plt.show()
```

```python
sns.countplot(x="Language", data=df)

plt.title("Count of Language")
plt.xlabel("Language")
plt.ylabel("Count")
plt.xticks(rotation=45)

plt.show()
```

```python
sns.pairplot(df, hue="Type")
plt.show()
```

```python
sns.violinplot(
    x="Type",
    y="Total_Stream",
    data=df,
    color="brown"
)

plt.title("Total Streams by Type")
plt.xlabel("Type")
plt.ylabel("Total Stream")

plt.show()
```

```python
sns.jointplot(
    x="Total_Stream",
    y="Solo_Streams_%",
    data=df,
    kind="hex"
)

plt.show()
```

```python
plt.figure(figsize=(6,4))

sns.boxplot(
    x="Type",
    y="Total_Stream",
    data=df
)

plt.title("Total Streams by Type")
plt.xlabel("Type")
plt.ylabel("Total Stream")

plt.show()
```
